## Execute the solution
The comments identify the operations and question parts.

```python
from pathlib import Path
import json
import pandas as pd

BASE = Path.cwd()
OUT = BASE / "outputs"
OUT.mkdir(exist_ok=True)
df = pd.read_excel(BASE / "data/student_records Question 4.xlsx",
                   sheet_name="Students", usecols="A:D")
# 4(a): two explicit representations of the same records.
records = [
    {"student_id": str(r["Student ID"]), "name": str(r["Name"]),
     "age": int(r["Age"]), "marks": int(r["Marks"])}
    for _, r in df.iterrows()
]
by_id = {r["student_id"]: (r["name"], r["age"], r["marks"])
         for r in records}
print("List of dictionaries sample:", records[0])
print("Dictionary of tuples sample:", {"S001": by_id["S001"]})


def read_records(path):
    try:
        with path.open("r", encoding="utf-8") as handle:
            return json.load(handle)
    except FileNotFoundError:
        print("Handled FileNotFoundError: records file was not found.")
        return None
    except json.JSONDecodeError:
        print("Handled JSONDecodeError: file is not valid JSON.")
        return None


def parse_marks(value):
    try:
        marks = int(value)
        if not 0 <= marks <= 100:
            raise ValueError("Marks must be between 0 and 100")
        return marks
    except (ValueError, TypeError):
        print("Handled invalid marks: enter a whole number from 0 to 100.")
        return None


# 4(b): write and then read the generated external file.
path = OUT / "student_records.json"
try:
    with path.open("w", encoding="utf-8") as handle:
        json.dump(records, handle, indent=2, ensure_ascii=False)
    print("Saved", len(records), "records to", path.name)
except OSError as error:
    print("Could not save records:", error)
else:
    retrieved = read_records(path)
    if retrieved is not None:
        print("Retrieved", len(retrieved), "records")
        print("Round-trip equality:", retrieved == records)
        for r in retrieved:
            print(f"{r['student_id']} | {r['name']} | "
                  f"Age {r['age']} | Marks {r['marks']}")
# 4(c): deliberately trigger two distinct errors.
# An empty temporary directory guarantees the file does not exist.
import tempfile
print("\nERROR TEST 1: missing file")
with tempfile.TemporaryDirectory() as temp:
    read_records(Path(temp) / "missing_students.json")
print("Continued after missing-file error.")
print("ERROR TEST 2: invalid numeric input 'seventy'")
parse_marks("seventy")
print("Continued after invalid-input error.")
print("Valid marks still accepted:", parse_marks("78"))
print("Program finished normally.")
```
