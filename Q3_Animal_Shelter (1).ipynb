## 3(a–c) Create the separate Python files
Expand the Colab Files sidebar to show both files. The strings below contain the complete source code.

```python
from pathlib import Path
Path("animals.py").write_text('"""Question 3: inheritance, encapsulation and polymorphism."""\n\nclass Animal:\n    def __init__(self, name, age, adopted=False):\n        self.name = name\n        self.set_age(age)\n        self.adopted = adopted\n\n    def get_age(self):\n        return self.__age\n\n    def set_age(self, age):\n        if isinstance(age, bool) or not isinstance(age, (int, float)):\n            raise TypeError("Age must be numeric.")\n        if age < 0:\n            raise ValueError("Age cannot be negative.")\n        self.__age = age\n\n    def adopt(self):\n        self.adopted = True\n\n    def display_info(self):\n        status = "Adopted" if self.adopted else "Available"\n        return f"{self.name}, age {self.get_age()}, {status}"\n\n    def make_sound(self):\n        return f"{self.name} makes an animal sound."\n\n\nclass Dog(Animal):\n    def __init__(self, name, age, breed, adopted=False):\n        super().__init__(name, age, adopted)\n        self.breed = breed\n\n    def make_sound(self):\n        return f"{self.name} says: Woof! Woof!"\n\n    def fetch(self):\n        return f"{self.name} is fetching the ball!"\n\n\nclass Cat(Animal):\n    def __init__(self, name, age, color, adopted=False):\n        super().__init__(name, age, adopted)\n        self.color = color\n\n    def make_sound(self):\n        return f"{self.name} says: Meow!"\n\n    def purr(self):\n        return f"{self.name} is purring."\n', encoding="utf-8")
Path("q3_main.py").write_text('from pathlib import Path\nimport pandas as pd\nimport animals\nfrom animals import Dog, Cat\n\nBASE = Path(__file__).resolve().parent\nrows = pd.read_excel(BASE / "data/animal_shelter question 3.xlsx",\n                     sheet_name="Animals", header=2, nrows=20,\n                     usecols="A:F")\nshelter = []\nfor _, row in rows.iterrows():\n    adopted = row["Adopted?"] == "Yes"\n    if row["Class"] == "Dog":\n        animal = Dog(row["Name"], int(row["Age (years)"]),\n                     row["Breed / Colour"], adopted)\n    elif row["Class"] == "Cat":\n        animal = Cat(row["Name"], int(row["Age (years)"]),\n                     row["Breed / Colour"], adopted)\n    else:\n        raise ValueError("Unknown animal class")\n    shelter.append(animal)\nprint("Imported module:", Path(animals.__file__).name)\nprint("Objects created:", len(shelter))\nprint("Adopted:", sum(a.adopted for a in shelter))\nprint("Available:", sum(not a.adopted for a in shelter))\n# The same calls dispatch to methods for different object types.\nfor animal in [shelter[0], shelter[10]]:\n    print(animal.display_info())\n    print(animal.make_sound())\nprint("Dog breed:", shelter[0].breed)\nprint("Cat colour:", shelter[10].color)\nprint(shelter[0].fetch())\nprint(shelter[10].purr())\nprint("Buddy age before:", shelter[0].get_age())\nshelter[0].set_age(4)\nprint("Buddy age after:", shelter[0].get_age())\nshelter[10].adopt()\nprint("After adoption:", shelter[10].display_info())\n# These demonstration changes affect memory, not the Excel source.\n', encoding="utf-8")
print("Created animals.py and q3_main.py")
```
